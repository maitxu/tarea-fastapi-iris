Hago la prueba del endpoint de health


```python
import requests

BASE_URL = "https://lung-earphone-backboned.ngrok-free.dev"

respuesta = requests.get(f"{BASE_URL}/health")
print(respuesta.status_code)
print(respuesta.json())
```

Hago la prueba del endpoint de predicción

```python
datos = {
    "sepal_length": 5.1,
    "sepal_width": 3.5,
    "petal_length": 1.4,
    "petal_width": 0.2
}

respuesta = requests.post(f"{BASE_URL}/predict", json=datos)
print(respuesta.status_code)
print(respuesta.json())
```
